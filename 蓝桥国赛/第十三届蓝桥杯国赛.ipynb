```python
import sys
from collections import deque

# 读取所有输入数据
input_data = sys.stdin.read().split()

if input_data:
    n = int(input_data[0])
    
    # 记录到达每个余数状态所需的最少 '1' 的个数
    # 初始化为正无穷大
    dist = [float('inf')] * n
    
    # 初始状态：数字 1
    # 它对 n 取余是 1 % n，此时有 1 个 '1'
    start = 1 % n
    dist[start] = 1
    
    # 使用双端队列进行 0-1 BFS
    q = deque([start])
    
    while q:
        u = q.popleft()
        
        # 如果余数为 0，说明找到了 n 的倍数，直接输出结果并结束整个程序
        if u == 0:
            print(dist[0])
            sys.exit() 
        
        # 选项 1：二进制末尾追加 0 (数值 * 2)
        # 不增加 '1' 的数量，边权为 0，加入队首
        v1 = (u * 2) % n
        if dist[u] < dist[v1]:
            dist[v1] = dist[u]
            q.appendleft(v1)
            
        # 选项 2：二进制末尾追加 1 (数值 * 2 + 1)
        # 增加了一个 '1'，边权为 1，加入队尾
        v2 = (u * 2 + 1) % n
        if dist[u] + 1 < dist[v2]:
            dist[v2] = dist[u] + 1
            q.append(v2)
```

```python
import sys
input = sys.stdin.readline

n, m, x = map(int, sys.stdin.readline().split())
a = list(map(int, sys.stdin.readline().split()))  # 每种书库存
books = [[] for _ in range(n)]  # 按书分类

max_day = 0
for _ in range(m):
    b, l, r = map(int, sys.stdin.readline().split())
    books[b-1].append((l, r))
    max_day = max(max_day, r)

ans = 0
extra_needed = []

for i in range(n):
    if not books[i]:
        continue
    # 处理每种书的预约
    intervals = sorted(books[i], key=lambda x: x[0])  # 按开始日排序
    # 差分数组
    diff = [0]*(max_day+3)
    for l, r in intervals:
        diff[l] += 1
        diff[r+1] -= 1
    prefix = [0]*(max_day+3)
    need = 0
    for day in range(1, max_day+2):
        prefix[day] = prefix[day-1] + diff[day]
        if prefix[day] > a[i]:
            need = max(need, prefix[day]-a[i])
    extra_needed.append((need, len(intervals)))  # 每种书的额外需求 + 预约数

# 贪心选择满足尽量多的预约
extra_needed.sort(key=lambda x: (x[0], -x[1]))  # 先需求少的，再预约多的
used_extra = 0
for need, total in extra_needed:
    if used_extra + need <= x:
        ans += total
        used_extra += need
    else:
        # 只能满足部分预约
        remaining = x - used_extra
        # 简化处理：假设每增加一本书可以多满足一条预约
        ans += min(total, remaining + a[0])  # a[0] 只是保底
        break

print(ans)
```
