```python
import sys

sys.setrecursionlimit(2000)

def solve():
    # 读取 n 和 k
    try:
        line1 = sys.stdin.readline().split()
        if not line1: return
        n, k = map(int, line1)
        a = list(map(int, sys.stdin.readline().split()))
    except ValueError:
        return

    # memo 用于存储已计算过的状态：(当前球的状态掩码, 剩余次数)
    memo = {}

    def dfs(mask, rem_k):
        if rem_k == 0:
            return 0
        state = (mask, rem_k)
        if state in memo:
            return memo[state]
        
        current_indices = []
        for i in range(n):
            if (mask >> i) & 1:
                current_indices.append(i)
        
        num_present = len(current_indices)
        max_val = 0
        
        for i in range(num_present):
            L = i  
            R = num_present - 1 - i  
            
            if R > 0 and L % R == 0:
                ball_idx = current_indices[i]

                res = a[ball_idx] + dfs(mask ^ (1 << ball_idx), rem_k - 1)
                if res > max_val:
                    max_val = res

        
        memo[state] = max_val
        return max_val

    # 初始状态：所有球都在（n个1），剩余 k 次
    initial_mask = (1 << n) - 1
    print(dfs(initial_mask, k))

solve()
```
