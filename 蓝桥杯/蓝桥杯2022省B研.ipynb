```python
# 俺的代码：
import os
import sys
data = list(input().strip())
i=0
n=len(data)-1
while i!=n:
  if (data[i] == '0' and data[i+1]=='0') or (data[i] == '1' and data[i+1]=='1'):
    data[i]='X' 
    del data[i+1]
    n-=1
  i+=1

for i in range(1,len(data)):
  if data[i]=='?':
    if data[i-1]=='0':
      data[i]='0'
    elif data[i-1]=='1':
      data[i]='1'
    else:
      if data[i+1]=='0':
        data[i]='0'
      else:
        data[i]='1'
i=0
n=len(data)-1
while i!=n:
  if (data[i] == '0' and data[i+1]=='0') or (data[i] == '1' and data[i+1]=='1'):
    data[i]='X' 
    del data[i+1]
    n-=1
  i+=1

print(data.count('X'))
```
